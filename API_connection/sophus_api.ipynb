```python
#saving data to csv
def save_to_csv(data, filename = "data.csv"):
    if isinstance(data, list):
        headers = data[0].keys()

        with open(filename, "w", newline="") as csv_file:
            writer = csv.DictWriter(csv_file, fieldnames=headers)
            writer.writeheader()
            writer.writerows(data)
        print(f"Data saved to {filename}")
    else:
        print("Data not suitable for CSV")

#Can change URL to other endpoints available in API doc
url = "https://api.sophus3.com/edx/digitalbrandperformance/brands/data?brand=volkswagen&limit=1000&datefrom=2024-01&dateto=2025-07" 

def get_and_save_data():
    access_token = get_access_token()

    headers = {
        'Authorization': f'Bearer {access_token}',
        's3-api-version': '2'
        }

    response = requests.get(url, headers=headers)
    print("Headers:", headers)

    if response.status_code == 200:
        print("API Data Retrieved Successfully")
        # print(response.text)
        data = response.json()

        if isinstance(data, dict) and "data" in data:
            save_to_csv(data["data"], "brand_data.csv")
        # elif "results" in data:
        #     save_to_csv(data['results'], "brand_data.csv")
        else:
            print("Unexpected data format", data)

    else:
        print(f"Request failed: {response.status_code} {response.text}")

get_and_save_data()
```
